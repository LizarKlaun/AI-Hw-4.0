```python
import numpy as np
```

```python
image_matrix = np.array([
    [100, 150, 200, 250],
    [50,  120, 180, 220],
    [30,   90, 140, 190],
    [10,   50, 100, 150]
])

darkened_image = np.clip(image_matrix - 20, 0, 255)

mean_intensity = np.mean(darkened_image)

print("Затемнене зображення:\n", darkened_image)
print(f"Середня інтенсивність: {mean_intensity}")
```

```python
coordinates = np.array([[1.5, 2.0], [5.0, 8.1], [3.2, 1.1], [8.9, 4.3], [2.1, 2.8]])

x, y = coordinates.T

max_x = np.max(x)
min_y = np.min(y)

mask = (x + y) > 6.0
filtered_coordinates = coordinates[mask]

print(f"X: {x}")
print(f"Y: {y}")
print(f"Max X: {max_x}, Min Y: {min_y}")
print("Точки із сумою > 6.0:\n", filtered_coordinates)
```

```python
stock_prices = np.array([100.0, 102.5, 101.0, 105.0, 104.2, 108.0, 110.5])

today_price = stock_prices[1:]
yesterday_price = stock_prices[:-1]

returns = (today_price - yesterday_price) / yesterday_price

best_day_idx = np.argmax(returns)

print("Щоденна дохідність:\n", returns)
print(f"Індекс дня з найбільшою дохідністю: {best_day_idx}")
print(f"Максимальна дохідність: {returns[best_day_idx]:.4%}")
```

```python
sensor_data = np.array([23.4, 24.1, -999, 25.0, 26.2, -999, 24.8])

missing_mask = (sensor_data == -999)

valid_mean = np.mean(sensor_data[~missing_mask])

cleaned_data = np.where(missing_mask, valid_mean, sensor_data)

print(f"Середнє валідних даних: {valid_mean:.2f}")
print("Дані після заміни пропусків:\n", cleaned_data)
```

```python
inventory = np.array([12, 45, 23, 67, 89, 34, 50])
target = 40

diff = np.abs(inventory - target)

closest_idx = np.argmin(diff)

closest_value = inventory[closest_idx]

print(f"Абсолютні різниці: {diff}")
print(f"Індекс найближчого значення: {closest_idx}")
print(f"Найближче значення в inventory: {closest_value}")
```
